```python
import requests

API_KEY = "SjPnudckMqvmnODB_qTgYCvUREE"
SELLER_ID = 71760
URL = "https://api-seller.uzum.uz/api/seller/shop/"

def token_check(api_key: str, seller_id: int) -> None:
    try:
        global r
        r = requests.get(
            URL,
            headers={"Authorization": f"Bearer {api_key}", "Accept": "application/json"},
            timeout=20,
        )
        r.raise_for_status()

        data = r.json()
        if not isinstance(data, list):
            print("false")
            print("store name: ")
            print("shop count: 0")
            return

        shop_count = len(data)
        shop = next((x for x in data if x.get("id") == seller_id), None)

        print("true")
        print(f"store name: {shop.get('shopTitle') if shop else 'NOT FOUND'}")
        print(f"shop count: {shop_count}")

    except requests.RequestException:
        print("false")
        print("store name: ")
        print("shop count: 0")

if __name__ == "__main__":
    token_check(API_KEY, SELLER_ID)
```
